```python
from collections import Counter
from src.data_schema import load_corpus, load_queries

docs = load_corpus(ROOT / 'data' / 'corpus.jsonl')
queries = load_queries(ROOT / 'data' / 'queries.jsonl')
print('Corpus', len(docs))
print('Queries', len(queries))
print('Split', Counter(q.split for q in queries))
print('Temporal', Counter(q.temporal_class for q in queries))
print('Needs=0', sum(q.needs_retrieval == 0 for q in queries))
print('Sample doc:', docs[0])
print('Sample query:', queries[0])
```
